```python
import sys
import pandas as pd
import polars as pl
import duckdb

print(sys.executable)
print("pandas:", pd.__version__)
print("polars:", pl.__version__)
print("duckdb:", duckdb.__version__)
```

```python
df = pl.DataFrame({
    "department": ["A", "A", "B", "B", "C"],
    "cases": [120, 135, 98, 110, 87]
})

df
```

```python
duckdb.sql("""
    SELECT
        department,
        SUM(cases) AS total_cases
    FROM df
    GROUP BY department
    ORDER BY total_cases DESC
""").show()
```
